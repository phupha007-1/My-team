```python
# ============================================
# PART 1 : IMPORT LIBRARIES
# ============================================

import random
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

from datetime import datetime, timedelta


# ใช้ seed เดิมทุกครั้งที่รัน
# เพื่อให้ข้อมูลสุ่มสามารถทำซ้ำได้
random.seed(7)


print("Import libraries สำเร็จ")
```

```python
# ============================================
# PART 2 : CLASS MEMBER
# ============================================

class Member:

    def __init__(self, member_id, name, is_vip=False):

        self.member_id = member_id
        self.name = name
        self.is_vip = is_vip


    def get_member_info(self):

        return {
            "member_id": self.member_id,
            "name": self.name,
            "is_vip": self.is_vip
        }


    def __str__(self):

        member_type = "VIP" if self.is_vip else "General"

        return (
            f"Member ID: {self.member_id} | "
            f"Name: {self.name} | "
            f"Type: {member_type}"
        )


print("Class Member พร้อมใช้งาน")
```

```python
# ============================================
# PART 3 : CLASS PACKAGE
# ============================================

class Package:

    def __init__(
        self,
        package_id,
        package_name,
        price,
        sessions,
        duration_days
    ):

        self.package_id = package_id
        self.package_name = package_name
        self.price = price
        self.sessions = sessions
        self.duration_days = duration_days


    def get_package_info(self):

        return {
            "package_id": self.package_id,
            "package_name": self.package_name,
            "price": self.price,
            "sessions": self.sessions,
            "duration_days": self.duration_days
        }


    def __str__(self):

        return (
            f"Package ID: {self.package_id} | "
            f"{self.package_name} | "
            f"{self.price:,.2f} บาท | "
            f"{self.sessions} sessions | "
            f"{self.duration_days} days"
        )


print("Class Package พร้อมใช้งาน")
```

```python
# ============================================
# PART 4 : CLASS GYMBOOKING
# ============================================

class GymBooking:

    # ราคาต่อชั่วโมงของแต่ละบริการ
    SERVICE_PRICES = {
        "Gym": 100,
        "Personal Training": 500,
        "Group Class": 250
    }


    def __init__(
        self,
        booking_id,
        member,
        package,
        hours,
        service_type,
        booking_date
    ):

        self.booking_id = booking_id
        self.member = member
        self.package = package
        self.hours = hours
        self.service_type = service_type
        self.booking_date = pd.to_datetime(booking_date)

        self.status = "Pending"


    def calculate_price(self):

        base_price = (
            self.SERVICE_PRICES
            .get(self.service_type, 0)
            * self.hours
        )

        # สมาชิก VIP ได้ส่วนลด 10%
        if self.member.is_vip:
            base_price *= 0.90

        return round(base_price, 2)


    def mark_done(self):

        self.status = "Done"


    def get_booking_info(self):

        return {
            "booking_id": self.booking_id,

            "member_id": self.member.member_id,
            "member_name": self.member.name,
            "is_vip": self.member.is_vip,

            "package_id": self.package.package_id,
            "package_name": self.package.package_name,

            "service_type": self.service_type,
            "hours": self.hours,

            "price": self.calculate_price(),

            "status": self.status,

            "booking_date": self.booking_date
        }


    def __str__(self):

        return (
            f"Booking ID: {self.booking_id} | "
            f"Member: {self.member.name} | "
            f"Package: {self.package.package_name} | "
            f"Service: {self.service_type} | "
            f"Price: {self.calculate_price():,.2f} บาท"
        )


print("Class GymBooking พร้อมใช้งาน")
```

```python
# ============================================
# PART 5 : GENERATE 300+ DATA
# ============================================


# --------------------------------------------
# รายชื่อสำหรับสุ่ม
# --------------------------------------------

first_names = [
    "Anan", "Somchai", "Somsak", "Niran",
    "Krit", "Thanakorn", "Weera", "Kitti",
    "Pim", "Nicha", "May", "Mint",
    "Fah", "Anna", "Jane", "Ploy",
    "Nok", "Aom", "Mali", "Beam"
]


last_names = [
    "Jaidee", "Dee", "Sukjai",
    "Thongdee", "Poomjai",
    "Wattanakul", "Saengdao",
    "Chaiyaporn", "Srisuk",
    "Kongkaew"
]


def generate_name():

    return (
        random.choice(first_names)
        + " "
        + random.choice(last_names)
    )


# --------------------------------------------
# สร้างวันที่สุ่มในปี 2026
# --------------------------------------------

def generate_booking_date():

    start_date = datetime(2026, 1, 1)

    total_days = 364

    random_days = random.randint(
        0,
        total_days
    )

    return (
        start_date
        + timedelta(days=random_days)
    )


# --------------------------------------------
# สร้าง Package
# --------------------------------------------

packages = [

    Package(
        1,
        "Basic",
        1200,
        10,
        30
    ),

    Package(
        2,
        "Premium",
        2500,
        20,
        30
    ),

    Package(
        3,
        "VIP",
        5000,
        50,
        60
    )
]


# --------------------------------------------
# ประเภทบริการ
# --------------------------------------------

services = [
    "Gym",
    "Personal Training",
    "Group Class"
]


# --------------------------------------------
# สร้างสมาชิก 300 คน
# --------------------------------------------

members_300 = []

member_data = {}


for i in range(1, 301):

    member = Member(
        member_id=i,
        name=generate_name(),
        is_vip=random.choice([True, False])
    )

    members_300.append(member)

    member_data[i] = member


# --------------------------------------------
# สร้าง Booking 300 รายการ
# --------------------------------------------

bookings_300 = []


for i in range(1, 301):

    selected_member = random.choice(
        members_300
    )

    selected_package = random.choice(
        packages
    )

    selected_service = random.choice(
        services
    )

    selected_hours = random.choice(
        [1, 1.5, 2, 2.5, 3]
    )

    booking = GymBooking(

        booking_id=i,

        member=selected_member,

        package=selected_package,

        hours=selected_hours,

        service_type=selected_service,

        booking_date=generate_booking_date()
    )

    booking.mark_done()

    bookings_300.append(
        booking
    )


print("สร้างข้อมูลสำเร็จ")
print("สมาชิก:", len(members_300))
print("Package:", len(packages))
print("Booking:", len(bookings_300))


print("\nตัวอย่าง Member")
print(members_300[0])


print("\nตัวอย่าง Package")
print(packages[0])


print("\nตัวอย่าง Booking")
print(bookings_300[0])
```

```python
# ============================================
# PART 6 : DATAFRAME + CSV
# ============================================


# --------------------------------------------
# แปลง Booking Object เป็น DataFrame
# --------------------------------------------

booking_rows = []


for booking in bookings_300:

    booking_rows.append(
        booking.get_booking_info()
    )


gym_df = pd.DataFrame(
    booking_rows
)


gym_df["booking_date"] = pd.to_datetime(
    gym_df["booking_date"]
)


# --------------------------------------------
# แปลง Member Object เป็น DataFrame
# --------------------------------------------

member_rows = []


for member in members_300:

    member_rows.append(
        member.get_member_info()
    )


member_df = pd.DataFrame(
    member_rows
)


# --------------------------------------------
# แปลง Package Object เป็น DataFrame
# --------------------------------------------

package_rows = []


for package in packages:

    package_rows.append(
        package.get_package_info()
    )


package_df = pd.DataFrame(
    package_rows
)


# --------------------------------------------
# บันทึก CSV
# --------------------------------------------

gym_df.to_csv(
    "fitness_gym_bookings.csv",
    index=False,
    encoding="utf-8-sig"
)


member_df.to_csv(
    "fitness_gym_members.csv",
    index=False,
    encoding="utf-8-sig"
)


package_df.to_csv(
    "fitness_gym_packages.csv",
    index=False,
    encoding="utf-8-sig"
)


print("สร้าง CSV สำเร็จ")

print(
    "Bookings:",
    len(gym_df)
)

print(
    "Members:",
    len(member_df)
)

print(
    "Packages:",
    len(package_df)
)


print("\nตัวอย่าง Booking DataFrame")
print(gym_df.head())
```

```python
# ============================================
# PART 7 : SQLITE DATABASE
# ============================================


DB_NAME = "fitness_gym.db"


conn = sqlite3.connect(
    DB_NAME
)


# เปิดใช้งาน Foreign Key
conn.execute(
    "PRAGMA foreign_keys = ON"
)


# --------------------------------------------
# สร้างตาราง Members
# --------------------------------------------

conn.execute(
    """
    CREATE TABLE IF NOT EXISTS members (

        member_id INTEGER PRIMARY KEY,

        name TEXT NOT NULL,

        is_vip INTEGER NOT NULL
    )
    """
)


# --------------------------------------------
# สร้างตาราง Packages
# --------------------------------------------

conn.execute(
    """
    CREATE TABLE IF NOT EXISTS packages (

        package_id INTEGER PRIMARY KEY,

        package_name TEXT NOT NULL,

        price REAL NOT NULL,

        sessions INTEGER NOT NULL,

        duration_days INTEGER NOT NULL
    )
    """
)


# --------------------------------------------
# สร้างตาราง Bookings
# --------------------------------------------

conn.execute(
    """
    CREATE TABLE IF NOT EXISTS bookings (

        booking_id INTEGER PRIMARY KEY,

        member_id INTEGER NOT NULL,

        package_id INTEGER NOT NULL,

        service_type TEXT NOT NULL,

        hours REAL NOT NULL,

        price REAL NOT NULL,

        status TEXT NOT NULL,

        booking_date TEXT NOT NULL,

        FOREIGN KEY (member_id)
            REFERENCES members(member_id),

        FOREIGN KEY (package_id)
            REFERENCES packages(package_id)
    )
    """
)


# --------------------------------------------
# ล้างข้อมูลเก่าก่อน Insert
# --------------------------------------------

conn.execute(
    "DELETE FROM bookings"
)

conn.execute(
    "DELETE FROM members"
)

conn.execute(
    "DELETE FROM packages"
)


# --------------------------------------------
# Insert Members
# --------------------------------------------

for _, row in member_df.iterrows():

    conn.execute(
        """
        INSERT INTO members
        (
            member_id,
            name,
            is_vip
        )
        VALUES (?, ?, ?)
        """,
        (
            int(row["member_id"]),
            row["name"],
            int(row["is_vip"])
        )
    )


# --------------------------------------------
# Insert Packages
# --------------------------------------------

for _, row in package_df.iterrows():

    conn.execute(
        """
        INSERT INTO packages
        (
            package_id,
            package_name,
            price,
            sessions,
            duration_days
        )
        VALUES (?, ?, ?, ?, ?)
        """,
        (
            int(row["package_id"]),
            row["package_name"],
            float(row["price"]),
            int(row["sessions"]),
            int(row["duration_days"])
        )
    )


# --------------------------------------------
# Insert Bookings
# --------------------------------------------

for _, row in gym_df.iterrows():

    conn.execute(
        """
        INSERT INTO bookings
        (
            booking_id,
            member_id,
            package_id,
            service_type,
            hours,
            price,
            status,
            booking_date
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?)
        """,
        (
            int(row["booking_id"]),
            int(row["member_id"]),
            int(row["package_id"]),
            row["service_type"],
            float(row["hours"]),
            float(row["price"]),
            row["status"],
            str(row["booking_date"])
        )
    )


conn.commit()


# --------------------------------------------
# ตรวจสอบจำนวนข้อมูล
# --------------------------------------------

for table_name in [
    "members",
    "packages",
    "bookings"
]:

    count = conn.execute(
        f"SELECT COUNT(*) FROM {table_name}"
    ).fetchone()[0]

    print(
        f"{table_name}: {count} rows"
    )


print("\nสร้าง SQLite Database สำเร็จ")
print(DB_NAME)
```

```python
# ============================================
# PART 8 : SQL SELECT / WHERE / ORDER BY
# ============================================


# --------------------------------------------
# เปิด Database
# --------------------------------------------

conn = sqlite3.connect(
    DB_NAME
)


# --------------------------------------------
# SELECT
# --------------------------------------------

query_select = """
SELECT
    booking_id,
    member_id,
    service_type,
    price
FROM bookings
"""


select_result = pd.read_sql_query(
    query_select,
    conn
)


print("===== SELECT =====")
print(select_result.head(10))


# --------------------------------------------
# WHERE + ORDER BY
# --------------------------------------------

query_where = """
SELECT
    booking_id,
    member_id,
    service_type,
    price
FROM bookings
WHERE price > 300
ORDER BY price DESC
"""


where_result = pd.read_sql_query(
    query_where,
    conn
)


print("\n===== WHERE + ORDER BY =====")
print(where_result.head(10))


# --------------------------------------------
# ตรวจสอบสมาชิก VIP
# --------------------------------------------

query_vip = """
SELECT
    member_id,
    name
FROM members
WHERE is_vip = 1
ORDER BY name
"""


vip_result = pd.read_sql_query(
    query_vip,
    conn
)


print("\n===== VIP MEMBERS =====")
print(vip_result.head(10))
```

```python
# ============================================
# PART 9 : GROUP BY + AGGREGATE + JOIN
# ============================================


# --------------------------------------------
# GROUP BY + COUNT + SUM
# --------------------------------------------

query_group = """
SELECT
    service_type,

    COUNT(*) AS total_bookings,

    SUM(hours) AS total_hours,

    SUM(price) AS total_revenue,

    AVG(price) AS average_price

FROM bookings

GROUP BY service_type

ORDER BY total_revenue DESC
"""


group_result = pd.read_sql_query(
    query_group,
    conn
)


print(
    "===== GROUP BY + AGGREGATE ====="
)

print(
    group_result
)


# --------------------------------------------
# JOIN 3 TABLES
# --------------------------------------------

query_join = """
SELECT

    b.booking_id,

    m.member_id,

    m.name AS member_name,

    p.package_id,

    p.package_name,

    p.price AS package_price,

    b.service_type,

    b.hours,

    b.price AS service_price,

    b.booking_date

FROM bookings AS b

JOIN members AS m

    ON b.member_id = m.member_id

JOIN packages AS p

    ON b.package_id = p.package_id

ORDER BY b.booking_date DESC
"""


join_result = pd.read_sql_query(
    query_join,
    conn
)


print(
    "\n===== JOIN 3 TABLES ====="
)

print(
    join_result.head(10)
)


# --------------------------------------------
# รายได้ตาม Package
# --------------------------------------------

query_package = """
SELECT

    p.package_name,

    COUNT(b.booking_id) AS total_bookings,

    SUM(b.price) AS total_revenue

FROM packages AS p

JOIN bookings AS b

    ON p.package_id = b.package_id

GROUP BY p.package_id, p.package_name

ORDER BY total_revenue DESC
"""


package_sql_result = pd.read_sql_query(
    query_package,
    conn
)


print(
    "\n===== REVENUE BY PACKAGE ====="
)

print(
    package_sql_result
)
```

```python
# ============================================
# PART 10 : PANDAS ANALYSIS
# ============================================


# --------------------------------------------
# GroupBy Service
# --------------------------------------------

service_analysis = (
    gym_df
    .groupby("service_type")
    .agg(
        total_bookings=(
            "booking_id",
            "count"
        ),

        total_hours=(
            "hours",
            "sum"
        ),

        total_revenue=(
            "price",
            "sum"
        ),

        average_price=(
            "price",
            "mean"
        )
    )
    .sort_values(
        "total_revenue",
        ascending=False
    )
)


print(
    "===== SERVICE ANALYSIS ====="
)

print(
    service_analysis
)


# --------------------------------------------
# GroupBy Package
# --------------------------------------------

package_analysis = (
    gym_df
    .groupby("package_name")
    .agg(
        total_bookings=(
            "booking_id",
            "count"
        ),

        total_hours=(
            "hours",
            "sum"
        ),

        total_revenue=(
            "price",
            "sum"
        ),

        average_price=(
            "price",
            "mean"
        )
    )
    .sort_values(
        "total_revenue",
        ascending=False
    )
)


print(
    "\n===== PACKAGE ANALYSIS ====="
)

print(
    package_analysis
)


# --------------------------------------------
# Top 10 สมาชิกที่ใช้เงินมากที่สุด
# --------------------------------------------

member_analysis = (
    gym_df
    .groupby(
        [
            "member_id",
            "member_name"
        ]
    )
    .agg(
        total_bookings=(
            "booking_id",
            "count"
        ),

        total_spent=(
            "price",
            "sum"
        )
    )
    .sort_values(
        "total_spent",
        ascending=False
    )
)


print(
    "\n===== TOP 10 MEMBERS ====="
)

print(
    member_analysis.head(10)
)
```

```python
# ============================================
# PART 11 : VISUALIZATION
# ============================================


# --------------------------------------------
# เตรียมข้อมูลรายเดือน
# --------------------------------------------

gym_df["month"] = (
    gym_df["booking_date"]
    .dt.to_period("M")
    .astype(str)
)


monthly_revenue = (
    gym_df
    .groupby("month")["price"]
    .sum()
)


# --------------------------------------------
# GRAPH 1
# จำนวน Booking ตามประเภทบริการ
# --------------------------------------------

service_count = (
    gym_df["service_type"]
    .value_counts()
)


plt.figure(
    figsize=(8, 5)
)

service_count.plot(
    kind="bar"
)

plt.title(
    "Number of Bookings by Service"
)

plt.xlabel(
    "Service Type"
)

plt.ylabel(
    "Number of Bookings"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# --------------------------------------------
# GRAPH 2
# รายได้ตาม Package
# --------------------------------------------

package_revenue = (
    gym_df
    .groupby("package_name")["price"]
    .sum()
    .sort_values(
        ascending=False
    )
)


plt.figure(
    figsize=(8, 5)
)

package_revenue.plot(
    kind="bar"
)

plt.title(
    "Revenue by Package"
)

plt.xlabel(
    "Package"
)

plt.ylabel(
    "Revenue (Baht)"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# --------------------------------------------
# GRAPH 3
# รายได้รายเดือน
# --------------------------------------------

plt.figure(
    figsize=(10, 5)
)

monthly_revenue.plot(
    kind="line",
    marker="o"
)

plt.title(
    "Monthly Revenue"
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Revenue (Baht)"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# --------------------------------------------
# สรุปผล
# --------------------------------------------

top_service = (
    service_count.idxmax()
)

top_package = (
    package_revenue.idxmax()
)

best_month = (
    monthly_revenue.idxmax()
)

best_month_revenue = (
    monthly_revenue.max()
)


print(
    "\n========== BUSINESS SUMMARY =========="
)

print(
    f"บริการที่มี Booking มากที่สุด: "
    f"{top_service}"
)

print(
    f"Package ที่สร้างรายได้สูงที่สุด: "
    f"{top_package}"
)

print(
    f"เดือนที่มีรายได้สูงที่สุด: "
    f"{best_month}"
)

print(
    f"รายได้ในเดือนดังกล่าว: "
    f"{best_month_revenue:,.2f} บาท"
)
```

```python
# ============================================
# PART 12 : ERROR HANDLING + REPORTS
# ============================================


# --------------------------------------------
# รายงานรายวัน
# --------------------------------------------

def daily_report():

    print(
        "\n========== DAILY REPORT =========="
    )


    date_input = input(
        "กรอกวันที่ (YYYY-MM-DD): "
    ).strip()


    try:

        selected_date = pd.to_datetime(
            date_input,
            format="%Y-%m-%d"
        )

    except ValueError:

        print(
            "รูปแบบวันที่ไม่ถูกต้อง"
        )

        print(
            "ตัวอย่าง: 2026-05-15"
        )

        return


    report = gym_df[
        gym_df["booking_date"].dt.date
        == selected_date.date()
    ]


    print(
        f"\nวันที่: {selected_date.date()}"
    )

    print(
        f"จำนวน Booking: {len(report)}"
    )

    print(
        f"รายได้รวม: "
        f"{report['price'].sum():,.2f} บาท"
    )


# --------------------------------------------
# รายงานรายสัปดาห์
# --------------------------------------------

def weekly_report():

    print(
        "\n========== WEEKLY REPORT =========="
    )


    date_input = input(
        "กรอกวันที่ในสัปดาห์ (YYYY-MM-DD): "
    ).strip()


    try:

        selected_date = pd.to_datetime(
            date_input,
            format="%Y-%m-%d"
        )

    except ValueError:

        print(
            "รูปแบบวันที่ไม่ถูกต้อง"
        )

        return


    start_date = (
        selected_date
        - pd.Timedelta(
            days=selected_date.weekday()
        )
    )


    end_date = (
        start_date
        + pd.Timedelta(days=6)
    )


    report = gym_df[
        (gym_df["booking_date"] >= start_date)
        &
        (
            gym_df["booking_date"]
            < end_date + pd.Timedelta(days=1)
        )
    ]


    print(
        f"\nช่วงวันที่ "
        f"{start_date.date()} "
        f"ถึง "
        f"{end_date.date()}"
    )

    print(
        f"จำนวน Booking: {len(report)}"
    )

    print(
        f"รายได้รวม: "
        f"{report['price'].sum():,.2f} บาท"
    )


# --------------------------------------------
# รายงานรายเดือน
# --------------------------------------------

def monthly_report():

    print(
        "\n========== MONTHLY REPORT =========="
    )


    try:

        year = int(
            input(
                "กรอกปี (YYYY): "
            )
        )

        month = int(
            input(
                "กรอกเดือน (1-12): "
            )
        )

    except ValueError:

        print(
            "กรุณากรอกปีและเดือนเป็นตัวเลข"
        )

        return


    if month < 1 or month > 12:

        print(
            "เดือนต้องอยู่ระหว่าง 1-12"
        )

        return


    report = gym_df[
        (gym_df["booking_date"].dt.year == year)
        &
        (gym_df["booking_date"].dt.month == month)
    ]


    print(
        f"\nรายงานเดือน "
        f"{month}/{year}"
    )

    print(
        f"จำนวน Booking: {len(report)}"
    )

    print(
        f"รายได้รวม: "
        f"{report['price'].sum():,.2f} บาท"
    )


# --------------------------------------------
# Report Menu
# --------------------------------------------

def report_menu():

    while True:

        print(
            "\n========== REPORT MENU =========="
        )

        print(
            "1. รายงานรายวัน"
        )

        print(
            "2. รายงานรายสัปดาห์"
        )

        print(
            "3. รายงานรายเดือน"
        )

        print(
            "0. กลับ"
        )


        choice = input(
            "เลือกเมนู: "
        ).strip()


        if choice == "1":

            daily_report()


        elif choice == "2":

            weekly_report()


        elif choice == "3":

            monthly_report()


        elif choice == "0":

            break


        else:

            print(
                "กรุณาเลือก 0 - 3"
            )
```

```python

# ============================================
# PART 13 : MAIN MENU
# ============================================


# --------------------------------------------
# อัปเดตข้อมูลหลังเพิ่มข้อมูลใหม่
# --------------------------------------------

def update_all_data():

    global gym_df
    global member_df
    global package_df


    # ----------------------------------------
    # Member DataFrame
    # ----------------------------------------

    member_rows = []

    for member in members_300:

        member_rows.append(
            member.get_member_info()
        )


    member_df = pd.DataFrame(
        member_rows
    )


    # ----------------------------------------
    # Package DataFrame
    # ----------------------------------------

    package_rows = []

    for package in packages:

        package_rows.append(
            package.get_package_info()
        )


    package_df = pd.DataFrame(
        package_rows
    )


    # ----------------------------------------
    # Booking DataFrame
    # ----------------------------------------

    booking_rows = []

    for booking in bookings_300:

        booking_rows.append(
            booking.get_booking_info()
        )


    gym_df = pd.DataFrame(
        booking_rows
    )


    gym_df["booking_date"] = pd.to_datetime(
        gym_df["booking_date"]
    )


    # ----------------------------------------
    # Save CSV
    # ----------------------------------------

    gym_df.to_csv(
        "fitness_gym_bookings.csv",
        index=False,
        encoding="utf-8-sig"
    )


    member_df.to_csv(
        "fitness_gym_members.csv",
        index=False,
        encoding="utf-8-sig"
    )


    package_df.to_csv(
        "fitness_gym_packages.csv",
        index=False,
        encoding="utf-8-sig"
    )


    # ----------------------------------------
    # Update SQLite
    # ----------------------------------------

    conn_update = sqlite3.connect(
        DB_NAME
    )


    conn_update.execute(
        "PRAGMA foreign_keys = ON"
    )


    conn_update.execute(
        "DELETE FROM bookings"
    )

    conn_update.execute(
        "DELETE FROM members"
    )

    conn_update.execute(
        "DELETE FROM packages"
    )


    # Members
    for _, row in member_df.iterrows():

        conn_update.execute(
            """
            INSERT INTO members
            VALUES (?, ?, ?)
            """,
            (
                int(row["member_id"]),
                row["name"],
                int(row["is_vip"])
            )
        )


    # Packages
    for _, row in package_df.iterrows():

        conn_update.execute(
            """
            INSERT INTO packages
            VALUES (?, ?, ?, ?, ?)
            """,
            (
                int(row["package_id"]),
                row["package_name"],
                float(row["price"]),
                int(row["sessions"]),
                int(row["duration_days"])
            )
        )


    # Bookings
    for _, row in gym_df.iterrows():

        conn_update.execute(
            """
            INSERT INTO bookings
            VALUES (?, ?, ?, ?, ?, ?, ?, ?)
            """,
            (
                int(row["booking_id"]),
                int(row["member_id"]),
                int(row["package_id"]),
                row["service_type"],
                float(row["hours"]),
                float(row["price"]),
                row["status"],
                str(row["booking_date"])
            )
        )


    conn_update.commit()

    conn_update.close()


# --------------------------------------------
# เพิ่มสมาชิก
# --------------------------------------------

def add_member():

    print(
        "\n========== ADD MEMBER =========="
    )


    name = input(
        "กรอกชื่อสมาชิก: "
    ).strip()


    if name == "":

        print(
            "ชื่อสมาชิกห้ามว่าง"
        )

        return


    vip_input = input(
        "เป็น VIP หรือไม่? (y/n): "
    ).strip().lower()


    if vip_input not in ["y", "n"]:

        print(
            "กรุณากรอก y หรือ n"
        )

        return


    is_vip = (
        vip_input == "y"
    )


    new_member_id = (
        max(member_data.keys()) + 1
    )


    new_member = Member(
        new_member_id,
        name,
        is_vip
    )


    members_300.append(
        new_member
    )


    member_data[
        new_member_id
    ] = new_member


    update_all_data()


    print(
        "\nเพิ่มสมาชิกสำเร็จ"
    )

    print(
        new_member
    )


# --------------------------------------------
# เพิ่ม Booking
# --------------------------------------------

def add_booking():

    print(
        "\n========== ADD BOOKING =========="
    )


    # Member ID
    try:

        member_id = int(
            input(
                "กรอก Member ID: "
            )
        )

    except ValueError:

        print(
            "Member ID ต้องเป็นตัวเลข"
        )

        return


    if member_id not in member_data:

        print(
            "ไม่พบสมาชิก"
        )

        return


    member = member_data[
        member_id
    ]


    # ----------------------------------------
    # แสดง Package
    # ----------------------------------------

    print(
        "\n========== PACKAGE =========="
    )


    for package in packages:

        print(
            f"{package.package_id}. "
            f"{package.package_name} - "
            f"{package.price:,.2f} บาท"
        )


    try:

        package_id = int(
            input(
                "เลือก Package ID: "
            )
        )

    except ValueError:

        print(
            "Package ID ต้องเป็นตัวเลข"
        )

        return


    selected_package = None


    for package in packages:

        if package.package_id == package_id:

            selected_package = package

            break


    if selected_package is None:

        print(
            "ไม่พบ Package"
        )

        return


    # ----------------------------------------
    # เลือกบริการ
    # ----------------------------------------

    print(
        "\n========== SERVICE =========="
    )

    print(
        "1. Gym"
    )

    print(
        "2. Personal Training"
    )

    print(
        "3. Group Class"
    )


    service_choice = input(
        "เลือกบริการ: "
    ).strip()


    service_map = {

        "1": "Gym",

        "2": "Personal Training",

        "3": "Group Class"
    }


    if service_choice not in service_map:

        print(
            "เลือกบริการไม่ถูกต้อง"
        )

        return


    service_type = service_map[
        service_choice
    ]


    # ----------------------------------------
    # จำนวนชั่วโมง
    # ----------------------------------------

    try:

        hours = float(
            input(
                "จำนวนชั่วโมง: "
            )
        )

    except ValueError:

        print(
            "จำนวนชั่วโมงต้องเป็นตัวเลข"
        )

        return


    if hours <= 0:

        print(
            "จำนวนชั่วโมงต้องมากกว่า 0"
        )

        return


    # ----------------------------------------
    # สร้าง Booking ID ใหม่
    # ----------------------------------------

    if len(bookings_300) == 0:

        new_booking_id = 1

    else:

        new_booking_id = (
            max(
                booking.booking_id
                for booking in bookings_300
            )
            + 1
        )


    # ----------------------------------------
    # สร้าง Booking Object
    # ----------------------------------------

    new_booking = GymBooking(

        booking_id=new_booking_id,

        member=member,

        package=selected_package,

        hours=hours,

        service_type=service_type,

        booking_date=datetime.now()
    )


    new_booking.mark_done()


    bookings_300.append(
        new_booking
    )


    # ----------------------------------------
    # Update CSV + SQLite
    # ----------------------------------------

    update_all_data()


    print(
        "\nเพิ่ม Booking สำเร็จ!"
    )

    print(
        new_booking
    )


# --------------------------------------------
# แสดงข้อมูลสมาชิก
# --------------------------------------------

def show_member():

    print(
        "\n========== MEMBER INFORMATION =========="
    )


    try:

        member_id = int(
            input(
                "กรอก Member ID: "
            )
        )

    except ValueError:

        print(
            "Member ID ต้องเป็นตัวเลข"
        )

        return


    if member_id not in member_data:

        print(
            "ไม่พบสมาชิก"
        )

        return


    member = member_data[
        member_id
    ]


    print(
        "\nข้อมูลสมาชิก"
    )

    print(
        member
    )


    member_bookings = []


    for booking in bookings_300:

        if (
            booking.member.member_id
            == member_id
        ):

            member_bookings.append(
                booking
            )


    print(
        "\nจำนวนครั้งที่ใช้บริการ:",
        len(member_bookings)
    )


    if len(member_bookings) == 0:

        print(
            "ยังไม่มีประวัติการใช้บริการ"
        )

        return


    total_spent = sum(

        booking.calculate_price()

        for booking in member_bookings
    )


    print(
        f"ยอดใช้จ่ายรวม: "
        f"{total_spent:,.2f} บาท"
    )


    print(
        "\nประวัติการใช้บริการ"
    )


    for booking in member_bookings:

        print(
            f"Booking ID: "
            f"{booking.booking_id} | "
            f"Package: "
            f"{booking.package.package_name} | "
            f"Service: "
            f"{booking.service_type} | "
            f"Hours: "
            f"{booking.hours} | "
            f"Price: "
            f"{booking.calculate_price():,.2f} บาท | "
            f"Date: "
            f"{booking.booking_date.date()}"
        )


# --------------------------------------------
# Main Menu
# --------------------------------------------

def main_menu():

    while True:

        print(
            "\n======================================"
        )

        print(
            "       FITNESS GYM MANAGEMENT"
        )

        print(
            "======================================"
        )

        print(
            "1. เพิ่มสมาชิก"
        )

        print(
            "2. เพิ่ม Booking / ใช้บริการ"
        )

        print(
            "3. แสดงข้อมูลสมาชิก"
        )

        print(
            "4. รายงาน"
        )

        print(
            "5. แสดง Package"
        )

        print(
            "6. ออกจากระบบ"
        )


        choice = input(
            "\nเลือกเมนู: "
        ).strip()


        # ------------------------------------
        # เพิ่มสมาชิก
        # ------------------------------------

        if choice == "1":

            add_member()


        # ------------------------------------
        # เพิ่ม Booking
        # ------------------------------------

        elif choice == "2":

            add_booking()


        # ------------------------------------
        # แสดงสมาชิก
        # ------------------------------------

        elif choice == "3":

            show_member()


        # ------------------------------------
        # รายงาน
        # ------------------------------------

        elif choice == "4":

            report_menu()


        # ------------------------------------
        # แสดง Package
        # ------------------------------------

        elif choice == "5":

            print(
                "\n========== AVAILABLE PACKAGES =========="
            )


            for package in packages:

                print(
                    package
                )


        # ------------------------------------
        # ออกจากระบบ
        # ------------------------------------

        elif choice == "6":

            print(
                "\nออกจากระบบเรียบร้อย"
            )

            print(
                "ขอบคุณที่ใช้ FITNESS GYM MANAGEMENT"
            )

            break


        else:

            print(
                "\nกรุณาเลือกเมนู 1 - 6"
            )


# ============================================
# START PROGRAM
# ============================================

main_menu()
```
